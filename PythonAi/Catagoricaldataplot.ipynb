Catagorical Data Plot
Now let`s discuss using seaborn to plot catagorical data ! There are few main plot types for this :
-->boxplot
-->violinplot
-->stripplot
-->swarmplot
-->barplot
-->countplot
Let`s go through examples of each!

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
df=sns.load_dataset('tips')
df
```

```python
sns.countplot(x=df['sex'],hue=df['smoker'])
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'])
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'],estimator=np.sum)  #this chech how much pay bill male and hoew much pay bill female
```

```python
sns.boxplot(x='tip',y='day',data=df)
```
